```python
import pandas as pd
```

```python
admission_df = pd.read_csv("../data/admission details.csv")
admission_df.columns = admission_df.columns.str.lower()
admission_df.rename(columns={
    "pt id":"pt_id",
    "doctor id":"doctor_id"
},inplace=True)

doctors_df = pd.read_csv("../data/doctors.csv")
doctors_df.columns = doctors_df.columns.str.lower()
doctors_df.rename(columns={
    "gender":"doctors_gender",
    "email":"doctors_email",
    "specialization":"doctors_specialization"
},inplace=True)

patient_df = pd.read_csv("../data/patient details.csv")
patient_df.columns = patient_df.columns.str.lower()
patient_df.rename(columns={
    "pt id":"pt_id",
    "name": "pt_name",
    "age": "pt_age",
    "sex": "pt_sex",
    "occupation": "pt_occupation",
    "level of education": "pt_level_of_education",
    "marital status": "pt_marital_status"
}, inplace=True)

risk_df = pd.read_csv("../data/risk factor.csv")
risk_df.columns = risk_df.columns.str.lower()
risk_df.rename(columns={"pt id":"pt_id"},inplace=True)

for df in [admission_df, doctors_df, patient_df, risk_df]:
    df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")

for df in [admission_df, doctors_df, patient_df, risk_df]:
    df[:] = df.apply(
        lambda col: col.str.strip().str.lower()
        if col.dtype == "str" else col
    )    
```

```python
admission_df = admission_df.merge(doctors_df,on="doctor_id",how="left")
admission_df = admission_df.merge(patient_df,on="pt_id",how="left")
admission_df = admission_df.merge(risk_df,on="pt_id",how="left")
admission_df.head()
```

```python
age_map = {
    **dict.fromkeys(range(0, 18), "0-17"),
    **dict.fromkeys(range(18, 30), "18-29"),
    **dict.fromkeys(range(30, 45), "30-44"),
    **dict.fromkeys(range(45, 60), "45-59"),
    **dict.fromkeys(range(60, 75), "60-74"),
    **dict.fromkeys(range(75, 101), "75+")
}

admission_df["age_range"] = admission_df["pt_age"].map(age_map)

group_map = {
    "0-17": "Under 18",
    "18-29": "Young adults",
    "30-44": "Adults",
    "45-59": "Middle-aged adults",
    "60-74": "Older adults",
    "75+": "Elderly adults"
}

admission_df["age_group"] = admission_df["age_range"].map(group_map)

admission_df[["pt_age", "age_range", "age_group"]].head(20)
```

```python
admission_map = {
    **dict.fromkeys(range(1, 4), "Short stay"),
    **dict.fromkeys(range(4, 8), "Moderate stay"),
    **dict.fromkeys(range(8, 15), "Extended stay"),
    **dict.fromkeys(range(15, 31), "Long stay"),
    **dict.fromkeys(range(31, 101), "Very long stay")
}

admission_df["admission_group"] = admission_df["admission_duration"].map(admission_map)

admission_df[["admission_duration", "admission_group"]].head(20)
```
